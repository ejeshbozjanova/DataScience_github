```python
y = True

number_in_my_mind = 55

while y:

    guess = int(input("Guess a number between 1 and 100."))

    if number_in_my_mind < guess:
        print(f"The number you entered: {guess}. Try a smaller number.")

    elif number_in_my_mind > guess:
        print(f"The number you entered: {guess}. Try a larger number.")

    else:
        print(f"The number you entered: {guess}. Correct guess. Congratulations!")
        y = False
```

```python
y = True

number_in_my_mind = 55

counter = 0

while y:

    guess = int(input("Guess a number between 1 and 100."))

    counter += 1

    if number_in_my_mind < guess:
        print(f"The number you entered: {guess}. Try a smaller number.")

    elif number_in_my_mind > guess:
        print(f"The number you entered: {guess}. Try a larger number.")

    else:
        print(f"The number you entered: {guess}. You guessed correctly in {counter} attempts. Congratulations!")
        y = False
```

```python
y = True

number_in_my_mind = 55

counter = 0

while y:

    guess = int(input("Guess a number between 1 and 100."))

    counter += 1

    if number_in_my_mind < guess:
        print(f"The number you entered: {guess}. Try a smaller number.")

    elif number_in_my_mind > guess:
        print(f"The number you entered: {guess}. Try a larger number.")

    else:
        print(f"The number you entered: {guess}. You guessed correctly in {counter} attempts. Congratulations!")
        y = False

    if counter == 5:
        print("You've reached 5 attempts. Game over.")
        y = False
```

```python
# Write a while loop that returns the square of the even elements in a list
numbers_list = [1, 3, 4, 6, 5, 7]

x = 0

even_squares_list = []

while x < len(numbers_list):

    if numbers_list[x] % 2 == 0:

        even_squares_list.append(numbers_list[x] ** 2)

    x += 1

print(even_squares_list)
```

```python
# Write a while loop that returns the square of odd elements and double of even elements in a list
numbers_list = [1, 5, 6, 7, 12, 14, 15, 18, 19, 32, 25]

x = 0

result_list = []

while x < len(numbers_list):

    if numbers_list[x] % 2 == 0:
        result_list.append(numbers_list[x] * 2)  # Double if the number is even
    else:
        result_list.append(numbers_list[x] ** 2)  # Square if the number is odd

    x += 1

print(result_list)
```
