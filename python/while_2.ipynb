[1,5,6,7,12,14,15,18,19,32,25] Write a while loop that returns a list where the square of the odd elements and twice the even elements of a given list are calculated. 

```python
numbers_list = [1,5,6,7,12,14,15,18,19,32,25] 

x = 0

result = []

while x < len(numbers_list):

    if numbers_list[x] % 2:
        result.append(numbers_list[x] ** 2)

    else:
        result.append(numbers_list[x] * 2)

    x += 1

print(result)
```

```python
numbers_list = [1,5,6,7,12,14,15,18,19,32,25] 

x = 0

result = []

while x < len(numbers_list):

    if numbers_list[x] % 2 == 1:
        result.append(numbers_list[x] ** 2)

    else:
        result.append(numbers_list[x] * 2)

    x += 1

print(result)
```

```python
import time
the_number_in_my_mind = random.randint(1,100)

time.sleep(2)

counter = 0

while True:

    guess = int(input("Guess a number between 1 and 100."))

    counter += 1

    time.sleep(2)

    if the_number_in_my_mind < guess:
        print(f"The number you entered. : {guess}. Decrease your guess.")

    elif the_number_in_my_mind > guess:
        print(f"The number you entered. : {guess}. Increase your guess")

    else:
        print(f"The number you entered. : {guess}. You guessed correctly. Congratulations! ")
        break

    time.sleep(2)

    if counter == 5:
        print(f"Your 5 chances are up. number {the_number_in_my_mind}")
        break
```

```python
list = [1, 2, 3, 4, "Techpro", 5, 6, 7] # Add the int values in order. When a str is encountered, stop the code.

x = 0

total = 0

while x < len(list):

    if type(list[x]) == int:
        total += list[x]

    elif type(list[x]) == str:
        break

    x += 1

print(total)
```

```python
list = [1, 2, 3, 4, "Techpro", 5, 6, 7] # Add the int values in order. When a string is encountered, skip that element and continue.

x = 0

total = 0

while x < len(list):

    if type(list[x]) == int:
        total += list[x]

    elif type(list[x]) == str:
        x += 1
        continue

    x += 1

print(total)
```

```python
list_1 = [4, 5, 6, 9, 85, 23, 65, 47, 32, 91]  # Find the smallest and largest elements in the list without using min, max, or sort functions.

smallest = list_1[0]
largest = list_1[0]

for number in list_1:
    if number < smallest:
        smallest = number
    elif number > largest:
        largest = number

print("Smallest:", smallest)
print("Largest:", largest)
```
